```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

```

```python
## drop month day year
df.drop(["day","month","year"], axis = 1 , inplace = True)
```

```python
df["Classes"].value_counts()
```

```python
# to solve it 

# encodig
df["Classes"] = np.where(df["Classes"].str.contains("not fire") , 0,1)
```

```python
df.head()
```

```python
df["Classes"].value_counts()
```

```python
# independe and dependent features
X = df.drop("FWI" , axis =1)

y = df["FWI"]
```

```python
X.head()
```

```python
y.head()
```

```python
# train test split
from sklearn.model_selection import train_test_split

X_train , X_test , y_train , y_test = train_test_split(X, y , test_size =0.25 , random_state = 42)
```

```python
X_train.shape , X_test.shape
```

```python
X_train.head()
```

```python
y_test.head()
```

```python
# feature selection based on correlation
# highly positive correlated feature canbe deleted
X_train.corr()
```

```python
#check for multicollinearity
plt.figure(figsize =(12,10))
corr = X_train.corr()
sns.heatmap(corr ,annot =True)
```

```python
def correlation(dataset , threshold):
    col_corr = set()
    corr_matrix  = dataset.corr()
    for i in range(len(corr_matrix.columns)):
        for j in range(i):
            if abs(corr_matrix.iloc[i , j]) > threshold:
                colname = corr_matrix.columns[i]
                col_corr.add(colname)
    return col_corr
```

```python
#set the threshold:  but it is set by domain expertise
corr_features = correlation(X_train , 0.85)
```

```python
corr_features
```

```python
X_train.drop(corr_features , axis=1 , inplace = True)
X_test.drop(corr_features , axis=1 , inplace = True)
```

# feature scaling or standardization


```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
X_train_scaled
```

## box plot to understande effect of standard scaler


```python
plt.subplots(figsize=(15, 5))
plt.subplot(1, 2, 1)
sns.boxplot(data=X_train)
plt.title('X_train Before Scaling')
plt.subplot(1, 2, 2)
sns.boxplot(data=X_train_scaled)
plt.title('X_train After Scaling')
```

## Linear Regression

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score

linreg = LinearRegression()

linreg.fit(X_train_scaled , y_train)
#predict
y_pred = linreg.predict(X_test)
mae = mean_absolute_error(y_test , y_pred)
score =  r2_score(y_test, y_pred)

print("mae : ",mae)
print(" r2 score: ",score)
```

```python
plt.scatter(y_test, y_pred) # it is following linear regression modelm so it is a good model
```

```python
# lets apply lasso regresion to overcome overfitting
```

# lasso regression

```python
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score

lasso = Lasso()

lasso.fit(X_train_scaled , y_train)
#predict
y_pred = lasso.predict(X_test)
mae = mean_absolute_error(y_test , y_pred)
score =  r2_score(y_test, y_pred)

print("mae : ",mae)
print("r2 score: ",score)
```

# Ridge regression model

```python
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score

ridge = Ridge()

ridge.fit(X_train_scaled , y_train)
#predict
y_pred = ridge.predict(X_test)
mae = mean_absolute_error(y_test , y_pred)
score =  r2_score(y_test, y_pred)

print("mae : ",mae)
print(" r2 score: ",score)
```

```python
# we ridgecv : model to do cross validation


from sklearn.linear_model import RidgeCV
ridgecv=RidgeCV(cv=5)
ridgecv.fit(X_train_scaled,y_train)
y_pred=ridgecv.predict(X_test_scaled)
plt.scatter(y_test,y_pred)
mae=mean_absolute_error(y_test,y_pred)
score=r2_score(y_test,y_pred)
print("Mean absolute error", mae)
print("R2 Score", score)
```

## elastic net

```python
from sklearn.linear_model import ElasticNet
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import r2_score

net= ElasticNet()

net.fit(X_train_scaled , y_train)
#predict
y_pred = net.predict(X_test)
mae = mean_absolute_error(y_test , y_pred)
score =  r2_score(y_test, y_pred)

print("mae : ",mae)
print(" r2 score: ",score)
```

```python
from sklearn.linear_model import ElasticNetCV
elasticcv=ElasticNetCV(cv=5)
elasticcv.fit(X_train_scaled,y_train)
y_pred=elasticcv.predict(X_test_scaled)
plt.scatter(y_test,y_pred)
mae=mean_absolute_error(y_test,y_pred)
score=r2_score(y_test,y_pred)
print("Mean absolute error", mae)
print("R2 Score", score)
```
