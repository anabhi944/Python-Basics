# Finding Elements

*  HTML content. The soup object contains all the data in the nested structure which could be programmatically extracted.

* all the content of the page is under the div with class entry-content.

* We will use the find class: 

This class will find the given tag with the given attribute. In our case, it will find all the div having class as entry-content

* you can see that all the images and links are also scraped. So our next task is to find only the content from the above-parsed HTML. On again inspecting the HTML of our website  

```python
# In the below example, we will find the elements by the class name
s = soup.find("div",class_ = "entry-content")

#content of the page is under the <p> tag.
content = s.find_all("p")

print(content)
```

# let’s see how to find elements by id.

* leftbar falls under the div tag with id as main



* list in the leftbar is under the <ul> tag with the class as leftBarList and our task is to find all the li under this ul.





```python
# finding by id 
s = soup.find("div",id ="main" )

#getting the leftbar 
leftbar = soup.find("ul" , class_ = "leftBarList")

# getting the li under the above ul
content = leftbar.find_all("li")

print(content)
```

# Extracting Text from the tags


* while scraping the data the tags also get scraped but what if we want only the text without any tags

* use text property


```python
s = soup.find('div', class_='entry-content') 
  
lines = s.find_all('p') 
  
for line in lines: 
    print(line.text)
```

# Removing the tags from the content of the leftbar


```python
# Finding by id 
s = soup.find('div', id= 'main') 
  
# Getting the leftbar 
leftbar = s.find('ul', class_='leftBarList') 
  
# All the li under the above ul 
lines = leftbar.find_all('li') 
  
for line in lines: 
    print(line.text)
```

# href attribute typically contains the URL that the anchor tag points to

```python
#we can see that images lie inside the img tag and the link of that image is inside the src attribute

# parsing the html 

soup = BeautifulSoup(r.content , "html.parser")

images_list = []

images = soup.select("img")
#This line uses the select method from BeautifulSoup to find all the <img> tags in the parsed HTML document (soup).



```

```python
# lets iterate throught each tag object i the images list .
# 
for image in images:
    src = image.get("src")  
    # retrieve the value of src attribute from "img" tag
    #src attribtue typically contain the url of image 
    
    alt = image.get("alt")
    # retrieve the value of the alt attribute from img tag
    # alt attribute typically contains alternative text for the image , 
    #which is usefull for accessibility purpose.
    
    images_list.append({"src": src , "alt":alt})
    # create a dictionary for each image 
    #src : image url and alt : image alternative text
    
    
```

```python
# print image

for image in images_list:
    print(image)
```

# dataframe 

```python
import requests
from bs4 import BeautifulSoup
```

```python
url = "https://www.webscraper.io/test-sites/e-commerce/allinone/computers/tablets"

r =  requests.get(url)

soup = BeautifulSoup(r.text , "lxml")

# we need to scrapp , name  ,price and ......

names = soup.find_all("a",class_ = "title")
print(names)
```

```python
# create empty list 

product_names = []

for name in names:
    names = name.text
    product_names.append(names)
    
print(product_names)
```

```python
# scrap price :

prices = soup.find_all("h4",class_ = "float-end price card-title pull-right")

price_list = []

for i in prices:
    price = i.text
    price_list.append(price)

print(price_list)
    
```

```python
# fetch  description 

desc = soup.find_all( "p" , class_ = "description card-text" )

desc_list  = []

for i in desc:
    descr = i.text
    desc_list.append(descr)
    
print(desc_list)
```

```python
# reviews 

reviews = soup.find_all("p", class_ = "float-end review-count")

rev = []

for i in reviews:
    re =  i.text
    rev.append(re)
    
print(rev)
```

```python
# create dataframe out of this.

import pandas as pd

df = pd.DataFrame({"product name": product_names,"prices":price_list,
                   "description" : desc_list})
df
```

# we nailed it 1  ouor first web scrapping.

```python
# to save this dataframe in excel sheet ,
df.to_csv("product_details.csv")  

```
