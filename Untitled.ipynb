```python
import numpy as np
from sklearn import datasets
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import LinearSVC
```

```python
iris = datasets.load_iris()
x = iris['data'][:,(2,3)]
y = (iris['target']==2).astype(np.float64())
```

```python
x
```

```python
y
```

```python
svm_clf = Pipeline([
    ('scaler',StandardScaler()),
    ('Linear_svc',LinearSVC(C=1,loss="hinge"))
])

svm_clf.fit(x,y)
```

```python
svm_clf.predict([[5.5,1.7]])
```

```python
# super simple with linear data , but what about the non linearity in dataset.
#add more features  to the data set , it will solve 

from sklearn.datasets import make_moons
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import train_test_split

x, y = make_moons(n_samples=100, noise=0.15)

x_train, x_test,y_train, y_test = train_test_split(x,y,test_size=0.25,random_state=42)

polynomial_svm_clf = Pipeline([
    ("poly_features",PolynomialFeatures(degree=5)),
    ("scaler",StandardScaler()),
    ("svm_clf",LinearSVC(C=1,loss="hinge"))
])

polynomial_svm_clf.fit(x,y)
```

polynomial feature methode : is computionally expensive once you increase degree , 

```python
# now lets  use kernal trick 
from sklearn.svm import SVC
poly_kernel_svm_clf = Pipeline([
            ("scaler",StandardScaler()),
            ("svm_clf",SVC(kernel="poly",degree = 3, coef0=1, C=5))
])
poly_kernel_svm_clf.fit(x,y)
```

```python
y_pred1 = poly_kernel_svm_clf.predict(x_test)
```

```python
precision_score(y_pred1,y_test)
```

```python
recall_score(y_pred1,y_test)   #better result than , polyfeature method
```

```python
confusion_matrix(y_pred1,y_test)
```

#similary feature  : another methode  to tackle non linear data 


```python
# similarity feature  , is expensive like polymial featuring so again 
# kernal trick do the magic 

rbf_kernel_svm = Pipeline([
    ("scaler",StandardScaler()),
    ("svm_clf",SVC(kernel="rbf",gamma=5,C=0.0001))
])

rbf_kernel_svm.fit(x,y)
```

```python
y_pred2 = rbf_kernel_svm.predict(x_test)
```

```python
print(recall_score(y_pred2,y_test))
print(precision_score(y_pred2,y_test))
print(confusion_matrix(y_pred2,y_test))
```
