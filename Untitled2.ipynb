```python
corpus = "Hello Welcome, to krish naik's NLP Tutorials. Please do watch the entire course!to become expert in NLP."
```

```python
print(corpus)
```

```python
## tokenization 
# sentense -> paragraph
from nltk.tokenize import sent_tokenize
```

```python
documents = sent_tokenize(corpus)
documents
```

```python
type(documents)
```

```python
for sentense in documents:
    print(sentense)
```

```python
#tokenization 
# paragraph-words
# sentenc- words
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentense in documents:
    print(word_tokenize(sentense))    
```

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus)           # making sure punctuation is also getting separeted , ' , s in naik's
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer = TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus) # it will not treat . (full stopp) as separate word
```
