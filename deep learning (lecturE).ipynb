```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

```python
model = Sequential()

model.add(Dense(3,activation = "tanh",input_dim = 11))
model.add(Dense(1,activation  = "tanh"))
```

* model = Sequential()

This line creates a new Sequential model. This model is essentially an empty container for adding layers to it one by one. Think of it as a linear pipeline where data will flow through each layer in sequence.

* model.add(Dense(3, activation="sigmoid", input_dim=11))

    * This line adds the first layer to the model. Here's what each argument  means:

    * Dense(3, ...) defines a dense (fully connected) layer with 3 neurons (units) in this layer. The number of neurons in this layer is a hyperparameter that you can adjust based on your specific problem.

    * activation="sigmoid" sets the activation function for the neurons in this layer to the sigmoid function. The sigmoid activation function squashes the output of each neuron into the range (0, 1).

   * input_dim=11 specifies that the input to this layer has 11 dimensions. This is typically used in the first layer of the model to define the input shape. In this case, it suggests that the input data should have 11 features.
  
  
*  model.add(Dense(1, activation="sigmoid"))
     * This line adds a second dense layer to the model. Here's what it does:

    * Dense(1, ...) defines a dense layer with a single neuron. This is often used as the output layer in binary classification problems, where you're trying to predict one of two classes.

     * activation="sigmoid" sets the activation function for this output neuron to sigmoid as well. This is common in binary classification tasks because it produces an output in the range (0, 1), which can be interpreted as a probability.
