```python
import pandas as pd
from bs4 import BeautifulSoup
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import cmudict

# Download necessary NLTK resources
nltk.download('punkt')
nltk.download('cmudict')

def extract_article_text(html_content):
    soup = BeautifulSoup(html_content, 'html.parser')

    # Extracting article title
    title = soup.find('title').get_text() if soup.find('title') else 'No Title'

    # Extracting article text
    article_text = ' '.join([p.get_text() for p in soup.find_all('p')])

    return title, article_text.strip()

'''def analyze_text(text):
    # Initialize SentimentIntensityAnalyzer
    sia = SentimentIntensityAnalyzer()

    # Tokenize the text into sentences and words
    sentences = sent_tokenize(text)
    words = word_tokenize(text)

    # Get CMU Pronouncing Dictionary for syllable count
    d = cmudict.dict()

    # Compute variables
    positive_score = sia.polarity_scores(text)['pos']
    negative_score = sia.polarity_scores(text)['neg']
    polarity_score = sia.polarity_scores(text)['compound']
    subjectivity_score = sia.polarity_scores(text)['compound']
    avg_sentence_length = len(words) / len(sentences)
    complex_word_count = len([word for word in words if len(word) > 2])
    percentage_of_complex_words = (complex_word_count / len(words)) * 100
    fog_index = 0.4 * (avg_sentence_length + percentage_of_complex_words)
    avg_words_per_sentence = len(words) / len(sentences)
    word_count = len(words)
    syllables_per_word = sum([len(d[word.lower()][0]) for word in words if word.lower() in d]) / len(words)
    personal_pronouns = sum(1 for word in words if word.lower() in ['i', 'me', 'my', 'mine', 'myself'])
    avg_word_length = sum(len(word) for word in words) / len(words)

    return [positive_score, negative_score, polarity_score, subjectivity_score, avg_sentence_length,
            percentage_of_complex_words, fog_index, avg_words_per_sentence, complex_word_count, word_count,
            syllables_per_word, personal_pronouns, avg_word_length]'''

'''def analyze_text(text):
    # Initialize SentimentIntensityAnalyzer
    sia = SentimentIntensityAnalyzer()

    # Tokenize the text into sentences and words
    sentences = sent_tokenize(text)
    words = word_tokenize(text)

    # Get CMU Pronouncing Dictionary for syllable count
    d = cmudict.dict()

    # Compute variables
    positive_score = sia.polarity_scores(text)['pos']
    negative_score = sia.polarity_scores(text)['neg']
    polarity_score = sia.polarity_scores(text)['compound']
    subjectivity_score = sia.polarity_scores(text)['neu']  # Fix subjectivity score calculation
    avg_sentence_length = len(words) / len(sentences)
    complex_word_count = len([word for word in words if len(word) > 2])
    percentage_of_complex_words = (complex_word_count / len(words)) * 100
    fog_index = 0.4 * (avg_sentence_length + percentage_of_complex_words)
    avg_words_per_sentence = len(words) / len(sentences)
    word_count = len(words)
    syllables_per_word = sum([len(d[word.lower()][0]) for word in words if word.lower() in d]) / len(words)
    personal_pronouns = sum(1 for word in words if word.lower() in ['i', 'me', 'my', 'mine', 'myself'])
    avg_word_length = sum(len(word) for word in words) / len(words)

    return [positive_score, negative_score, polarity_score, subjectivity_score, avg_sentence_length,
            percentage_of_complex_words, fog_index, avg_words_per_sentence, complex_word_count, word_count,
            syllables_per_word, personal_pronouns, avg_word_length] '''

'''def analyze_text(text):
    # Initialize SentimentIntensityAnalyzer
    sia = SentimentIntensityAnalyzer()

    # Tokenize the text into sentences and words
    sentences = sent_tokenize(text)
    words = word_tokenize(text)

    # Get CMU Pronouncing Dictionary for syllable count
    d = cmudict.dict()

    # Compute variables
    positive_score = sia.polarity_scores(text)['pos']
    negative_score = sia.polarity_scores(text)['neg']
    polarity_score = sia.polarity_scores(text)['compound']
    subjectivity_score = sia.polarity_scores(text)['neu']  # Fix subjectivity score calculation
    avg_sentence_length = len(words) / len(sentences) if len(sentences) > 0 else 0  # Check for division by zero
    complex_word_count = len([word for word in words if len(word) > 2])
    percentage_of_complex_words = (complex_word_count / len(words)) * 100
    fog_index = 0.4 * (avg_sentence_length + percentage_of_complex_words)
    avg_words_per_sentence = len(words) / len(sentences) if len(sentences) > 0 else 0  # Check for division by zero
    word_count = len(words)
    syllables_per_word = sum([len(d[word.lower()][0]) for word in words if word.lower() in d]) / len(words)
    personal_pronouns = sum(1 for word in words if word.lower() in ['i', 'me', 'my', 'mine', 'myself'])
    avg_word_length = sum(len(word) for word in words) / len(words)

    return [positive_score, negative_score, polarity_score, subjectivity_score, avg_sentence_length,
            percentage_of_complex_words, fog_index, avg_words_per_sentence, complex_word_count, word_count,
            syllables_per_word, personal_pronouns, avg_word_length]'''

def analyze_text(text):
    # Initialize SentimentIntensityAnalyzer
    sia = SentimentIntensityAnalyzer()

    # Tokenize the text into sentences and words
    sentences = sent_tokenize(text)
    words = word_tokenize(text)

    # Get CMU Pronouncing Dictionary for syllable count
    d = cmudict.dict()

    # Compute variables
    positive_score = sia.polarity_scores(text)['pos']
    negative_score = sia.polarity_scores(text)['neg']
    polarity_score = sia.polarity_scores(text)['compound']
    subjectivity_score = sia.polarity_scores(text)['neu']  # Fix subjectivity score calculation
    avg_sentence_length = len(words) / len(sentences) if len(sentences) > 0 else 0  # Check for division by zero
    complex_word_count = len([word for word in words if len(word) > 2])
    
    # Check for division by zero when calculating percentage_of_complex_words
    percentage_of_complex_words = (complex_word_count / len(words)) * 100 if len(words) > 0 else 0
    
    fog_index = 0.4 * (avg_sentence_length + percentage_of_complex_words)
    avg_words_per_sentence = len(words) / len(sentences) if len(sentences) > 0 else 0  # Check for division by zero
    word_count = len(words)
    syllables_per_word = sum([len(d[word.lower()][0]) for word in words if word.lower() in d]) / len(words)
    personal_pronouns = sum(1 for word in words if word.lower() in ['i', 'me', 'my', 'mine', 'myself'])
    avg_word_length = sum(len(word) for word in words) / len(words)

    return [positive_score, negative_score, polarity_score, subjectivity_score, avg_sentence_length,
            percentage_of_complex_words, fog_index, avg_words_per_sentence, complex_word_count, word_count,
            syllables_per_word, personal_pronouns, avg_word_length]




def main():
    # Load the Excel file into a DataFrame
    input_file = 'input.xlsx'
    df = pd.read_excel(input_file)

    # Create a DataFrame for output structure
    output_columns = ["URL_ID"] + ["VAR" + str(i) for i in range(1, 14)]
    output_df = pd.DataFrame(columns=output_columns)

    for index, row in df.iterrows():
        url_id = row['URL_ID']
        html_content = row['URL']  # Replace 'HTML_CONTENT_COLUMN_NAME' with the actual column name containing HTML content

        # Extract article title and text
        _, article_text = extract_article_text(html_content)

        # Analyze text and compute variables
        variables = [url_id] + analyze_text(article_text)

        # Append to output DataFrame
        output_df = output_df.append(dict(zip(output_columns, variables)), ignore_index=True)

    # Save the output DataFrame to a new Excel file
    output_file = 'output_structure_results.xlsx'
    output_df.to_excel(output_file, index=False)

if __name__ == "__main__":
    main()
```
