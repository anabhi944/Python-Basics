```python
# tokenization

from spacy.lang.en import English

# laod english tokenizer , taggar , parser , NER , word and vectors 

nlp = English()

text = """ when learning data science you should not get discouraged! challenges and setbacks are not failures ,
they are just part of journey"""

# "nlp" object is used to create documents with lunguistic annotation.
my_doc = nlp(text)
```

```python
#create list of word tokens 
token_list = []

for token in  my_doc:
    token_list.append(token.text)
    
print(token_list)
```

```python
# sentence tokenization

# laod english tokenizer , tagger , parse , NER , and word vectors 
nlp = English()

# add the sentencizer component to the pipline 
nlp.add_pipe("sentencizer")

text = """hello i am anabhi . how are you ? I know you do not know me , but I know you very well . you are new to collage . dont worry .I will help you out."""

# nlp object is used to create documnents with linguistic annotation 

doc = nlp(text)

# create lsit of sentence tokens

sents_list = []

for sent in doc.sents:
    sents_list.append(sent.text)

print(sents_list)
```
