```python
import phonenumbers
from phonenumbers import geocoder
from geopy.geocoders import Nominatim

def get_location(phone_number):
    try:
        # Parse the phone number and extract the country code
        parsed_number = phonenumbers.parse(phone_number, None)
        country_code = phonenumbers.region_code_for_country_code(parsed_number.country_code)
        country_name = phonenumbers.region_name_for_country_code(parsed_number.country_code)

        # Use geocoder to get detailed location information
        geolocator = Nominatim(user_agent="phone_location_app")
        location = geolocator.geocode(country_name)
        if location:
            return {
                "country": location.address,
                "state": geocoder.description_for_number(parsed_number, "en"),
                "city": location.raw.get('address', {}).get('city', 'N/A'),
                "street": location.raw.get('address', {}).get('road', 'N/A')
            }
        else:
            return "Location not found"
    except phonenumbers.phonenumberutil.NumberParseException:
        return "Invalid phone number format"

# Example usage
phone_number = "+916266544765"  # Replace with the phone number you want to check
location = get_location(phone_number)
print("Location:", location)
```
