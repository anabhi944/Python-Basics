```python
import pandas as pd
from bs4 import BeautifulSoup
import requests

# Load links.csv to get IMDB IDs
links_df = pd.read_csv("C:\\Users\\lenovo\\Downloads\\movie_data\\links.csv")

# Assuming the 'imdbId' column contains IMDB IDs
imdb_ids = links_df['imdbId'].astype(str)

# Function to scrape IMDB rating from a given IMDB ID
def scrape_imdb_rating(imdb_id):
    url = f'https://www.imdb.com/title/{imdb_id}/'
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')
    rating_element = soup.find('span', {'itemprop': 'ratingValue'})
    return float(rating_element.text) if rating_element else None

# Add IMDB ratings to links_df
links_df['imdb_rating'] = imdb_ids.apply(scrape_imdb_rating)

# Task 1: Movie with the highest IMDB rating
highest_rated_movie = links_df.loc[links_df['imdb_rating'].idxmax(), 'movieId']

# Task 2: Sci-Fi movie with the highest IMDB rating
sci_fi_movies = links_df[links_df['genres'].str.contains('Sci-Fi')]
highest_rated_sci_fi_movie = sci_fi_movies.loc[sci_fi_movies['imdb_rating'].idxmax(), 'movieId']

print(f"Movie with the highest IMDB rating (Task 1): {highest_rated_movie}")
print(f"Sci-Fi movie with the highest IMDB rating (Task 2): {highest_rated_sci_fi_movie}")
```
