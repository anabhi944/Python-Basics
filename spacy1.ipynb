# Tokenization

spaCy is designed to make it easy to build systems for information extraction or general-purpose natural language processing

```python
# word tokenization
from spacy.lang.en import English

# Load english tokenizer , tagger , parser , NER and word vectors 
nlp = English()


text = """When learning data science, you shouldn't get discouraged!
Challenges and setbacks aren't failures, they're just part of the journey. You've got this!"""

# "nlp" object is used to create documents with linguistic annotations
my_doc = nlp(text)
```

```python
# create list of words token
token_list = []

for token in my_doc:
    token_list.append(token.text)
    
print(token_list)
```

# sentence tokenization

```python

# loas english tokenizer , tagger , parser , NER , and word vector
nlp = English() 

# add the " sentencizer" component to the pipeline 
nlp.add_pipe("sentencizer")

text = """when learning data science, you should't get discouraged! challenges and setbacks aren't failures, they are just part of our journey. You've got this!"""

# nlp object is used to create documents with linguistic annptaions.
doc = nlp(text)

# create list of sentence tokens 
sents_list = []

for sent in doc.sents:
    sents_list.append(sent.text)
    
print(sents_list)
```
