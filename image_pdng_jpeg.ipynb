# Image Formation : 

**generating synthetic image , camera calibration and 3d reconstruction**

```python
import numpy as np
import cv2
```

```python
# generat a synthetic image ( white canvas) 
width , height = 400 ,300
synthetic_image = np.ones((height, width ,3) , dtype = np.uint8)* 255

```

```python
# draw a red rectangle on th image 
start_point = (100 , 100)
end_point = (300,200)

color = (0, 0, 255)  # red color in BGR format
thickness = 2
cv2.rectangle(synthetic_image, start_point, end_point, color, thickness)
```

```python
# display rectangle
cv2.imshow("image", synthetic_image)
cv2.waitKey(0)
```

# image segmentation : dividing image into meanigful regions or objects

```python
import cv2
```

```python
# read image 
image_path = "C:\\Users\\lenovo\\Pictures\\amanclicks\\alok.jpg"
image = cv2.imread(image_path)

```

```python
# convert the imgae to grayscale

gray__image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

```

Image thresholding is a common image processing technique used to separate objects of interest from the background in a digital image. It's particularly useful when you have a grayscale image or a single-channel image (e.g., a black and white image) and you want to create a binary image where pixels are classified as either part of the object or part of the background.

```python
#apply image thresholding to segment objects form the background
_, binary_image = cv2.threshold(gray_image, 128, 255, cv2.THRESH_BINARY)

```

* 128: This is the threshold value. Pixels in gray_image with intensity values greater than or equal to 128 will be set to 255 (white) in the resulting binary image, while pixels with values less than 128 will be set to 0 (black).

* 255: This is the maximum value to which the pixels are set when they meet the threshold condition. In this case, it's 255, which represents white in an 8-bit image.

* cv2.THRESH_BINARY: This is the thresholding method. cv2.THRESH_BINARY is a binary thresholding method, which means pixels are set to one of two values (0 or 255) based on the threshold condition.

* cv2.threshold returns two values: a retval (which contains the chosen threshold) and the binary image. Since you're not interested in the retval, you use _ 

```python
# display the originaal and segmented imAGES
cv2.imshow("image", image)
cv2.imshow("image",binary_image)
cv2.waitkey(0)
```
