```python
import pandas as pd
import numpy as np
from sklearn import linear_model
import matplotlib.pyplot as plt
```

```python
df =  pd.read_excel("C:\\Users\\lenovo\\Documents\\homeprices.xlsx")
```

```python
df
```

```python
%matplotlib inline
plt.xlabel("area")
plt.ylabel("price")
plt.scatter(df.area , df.price , color = "red",marker = "+")
```

```python
new_df = df.drop("price" , axis = "columns")
```

```python
new_df
```

```python
price = df.price
price
```

```python
# create linear regresion object
reg = linear_model.LinearRegression()
reg.fit(new_df , price)
```

# 1. predict price of a home with area  = 3300 sqr fit

```python
reg.predict([[3300]])
```

```python
reg.coef_                      # b
```

```python
reg.intercept_                 #m
```

```python
# firstly import dataset 
canada_df = pd.read_csv("C:\\Users\\lenovo\\Downloads\\canada_per_capita_income.csv")
```

```python
canada_df
```

```python
canada_df.rename(columns = {"per capita income (US$)" : "per_capita_income"}, inplace =True)
```

```python
canada_df.head(6)
```

```python
%matplotlib inline
plt.xlabel("year")
plt.ylabel("per_capita_income")
plt.scatter(canada_df.year , canada_df.per_capita_income, color ="red", marker = "+")
```

```python
new_df = canada_df.drop("per_capita_income" , axis ="columns")
new_df.head(5)
```

```python
per_capita_income = canada_df.per_capita_income
per_capita_income.head(5)
```

```python
# make object of linear regression
reg = linear_model.LinearRegression()
reg.fit(new_df , per_capita_income)
```

```python
reg.predict([[2020]])
```
