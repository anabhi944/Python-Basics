```python
# Extract the relevant columns for clustering
X = mall_clusters.iloc[:, [3, 4]].values  # We use Annual Income and Spending Score for clustering


from sklearn.cluster import KMeans

# Initialize an empty list to store the within
```

```python
# Initialize an empty list to store the within-cluster sum of squares (WCSS) values
wcss = []           

```

```python
# Calculate WCSS for different values of k (number of clusters)
for i in range(1, 11):
    km = KMeans(n_clusters = i )
    km.fit_predict(X)
    wcss.append(km.inertia_)


```

```python
wcss
```

```python
plt.plot(range(1,11),wcss)
```

```python
# extract numpy array 
#x

# make a new object (Km) 
km = KMeans(n_clusters =4)
y_means = km.fit_predict(X)
```

```python
# Plot the Elbow method graph
plt.figure(figsize=(10,5))
plt.plot(range(1, 11), wcss, marker='o', linestyle='--')
plt.title('Elbow Method')
plt.xlabel('Number of clusters (k)')
plt.ylabel('WCSS')
plt.show()
```
