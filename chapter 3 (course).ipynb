# our own classification dataset

**Precision** is like being very careful. It tells us how many of the things we said are true are actually true. So, if we say, "This is positive," how often are we correct? High precision means we're really good at being right when we say something is positive.

**Recall** is like not missing things. It tells us how many of the true positive things we found out of all the positive things that exist. So, if there are 100 positive things, how many did we manage to find? High recall means we're good at finding most of the positive things.

In simple terms, precision is about avoiding mistakes, while recall is about not missing important stuff. Depending on the situation, we might care more about one or the other. Sometimes we want to be really sure when we say something is positive (high precision), and other times we want to make sure we find as many positives as possible (high recall). It's a balance between being cautious and thorough.

**The harmonic mean,**
often denoted as "F1 score," is commonly used in classification problems to provide a balanced measure of a model's performance, especially when dealing with imbalanced datasets or when both precision and recall are important. Here are some scenarios where the harmonic mean (F1 score) is used in classification problems:

**Imbalanced Datasets:**
When you have a classification problem where one class greatly outnumbers the other (class imbalance), accuracy can be misleading. The F1 score is useful in such cases because it considers both false positives (precision) and false negatives (recall). It provides a single metric that balances these two aspects of classification performance.

**Information Retrieval and Text Classification:** In information retrieval and text classification tasks, the F1 score is often used to evaluate the effectiveness of search engines or text classifiers. Precision represents the relevance of the retrieved documents, while recall measures how many relevant documents were found.

**Medical Diagnosis:** In medical diagnosis, correctly identifying both positive and negative cases is crucial. The F1 score is used to strike a balance between minimizing false positives and false negatives, ensuring that the model is both accurate and sensitive to detecting diseases or conditions.

**Anomaly Detection:** In anomaly detection problems, where the goal is to identify rare events or anomalies, the F1 score is used to assess the performance of the model. Achieving a high F1 score implies a good balance between correctly identifying anomalies (recall) and avoiding false alarms (precision).

**Spam Detection:** When building spam email filters, it's essential to avoid false positives (misclassifying legitimate emails as spam) while catching as much spam as possible. The F1 score helps find the right balance between precision and recall.

**Fraud Detection:**
Detecting fraudulent transactions in financial data requires both precision (to avoid blocking legitimate transactions) and recall (to catch as many fraudulent transactions as possible). The F1 score helps evaluate the model's overall performance in this context.

 * In these and similar scenarios, the harmonic mean (F1 score) is used because it provides a single metric that takes into account both precision and recall, helping practitioners assess and compare classification models effectively.
 * It allows you to make informed decisions about model tuning, threshold selection, and trade-offs between false positives and false negatives based on the specific requirements of the problem.

# GRID SEARCH : is hyperparamter optimization

* **Grid search** is a technique used in machine learning to systematically search for the optimal combination of hyperparameters for a given machine learning algorithm. 

* TO find the best set of features .

* **Hyperparameters** are settings or configurations that are not learned from the data but are set prior to training a model.

* These hyperparameters can significantly impact the performance of a machine learning model, and finding the right combination can be a challenging task.


**grid search performed on only traning set , bcz we dont want to tune your hyperparameters based in the test.**

**define based model (accordingly , classification/regression model) before jumping into the gridsearchcv**

**step** : to make a search space 

set of values of the hyperparamters that you want to search in order to find the best set of values for hyper paramaters.

make dictionary.

n_estimaters: no. of tree in the forest

max_features : no. of feature to conside when looking for the best split 

bootstrap: Whether bootstrap samples are used when building trees.

n_estimators: Number of trees in the forest.

max_features: The number of features to consider when looking for the best split.





```python
#lets combine all measuremnet in single function 

def algo_accuracy(y_test , predictions):
    conf_matrix = confusion_matrix(y_test , predictions)
    accuracy = accuracy_score(y_test , predictions)
    precision = precision_score(y_test , predictions)
    recall = recall_score(y_test , predictions)
    
    return conf_matrix , accuracy , precision , recall
```

```python
# lets do grid search :

from sklearn.model_selection import GridSearchCV

param_grid = [
    
    # try 12 (3x4) combination of hyperparameters
    {"n_estimators": [3,10,30] , "max_features":[2,4,6,8]},
    
    #then try 6 (2x3) combination with bootstrap set as False
    {"bootstrap": [False] ,"n_estimators": [3,10] , "max_features": [2,3,4]},
    
]

#random_forest 
from sklearn.ensemble import RandomForestClassifier
rfc = RandomForestClassifier()

#train acros 5 folds , thats total of (12+6)*5 = 90 rounds of traning 

grid_search = GridSearchCV(rfc , param_grid , cv = 5 , 
                          scoring = "accuracy",
                          return_train_score = True)

grid_search.fit(scaled_X_train , y_train)
```

```python
rfc_clf = grid_search.best_estimator_
```

```python
# when maximum featue is 2 , and estimator is 30 , it best predict .
```

```python
rfc_predictions = rfc_clf.predict(scaled_X_test)
rfc_predictions
```

# types of classification

**1 .Multiclass Classification:**

*Definition:*
In multiclass classification, a single data point is assigned to one and only one class out of several possible classes. Each class is mutually exclusive, meaning that a data point cannot belong to more than one class simultaneously.

*Example:*
Digit recognition, where you classify handwritten digits (0-9) into one of the ten possible classes.

**2.Multilabel Classification:**

*Definition:*

In multilabel classification, each data point can be associated with multiple classes or labels. This means that a data point may belong to zero, one, or more than one class simultaneously. It allows for overlapping classes.

*Example:*

Classifying news articles into multiple categories or topics. An article can be labeled as both "politics" and "economics" if it covers aspects of both subjects.

**3.Multioutput Classification:**

*Definition:*

Multioutput classification refers to a scenario where each data point is associated with multiple output variables, and each of these output variables can take on multiple classes or labels. It's a generalization of both multiclass and multilabel classification.

*Example:*

Predicting the attributes of an image, such as classifying objects in the image (multiclass) and tagging it with various attributes (multilabel). For example, in an image of a dog, it can be classified as a "dog" (multiclass) and tagged with attributes like "fluffy," "brown," "playful," etc. (multilabel).

Here's a summary of the key differences:

* Multiclass: Each data point belongs to exactly one class out of several mutually exclusive classes.

* Multilabel: Each data point can belong to multiple classes simultaneously, and classes are not mutually exclusive.

* Multioutput: Each data point is associated with multiple output variables, and each of these variables can take on multiple classes or labels.

The choice of which type of classification to use depends on the nature of the problem and the characteristics of the data. Different algorithms and techniques may be employed for each type of classification task.
